```python
!pip install datasets

import pandas as pd
from datasets import load_dataset
from sklearn.model_selection import train_test_split
```

# 📰 🟢 CELL 2 — Load AG News Dataset (Auto)

```python
ag_dataset = load_dataset("ag_news")

train_ag = pd.DataFrame(ag_dataset['train'])
test_ag = pd.DataFrame(ag_dataset['test'])

ag_df = pd.concat([train_ag, test_ag], ignore_index=True)

ag_label_map = {
    0: "World",
    1: "Sports",
    2: "Business",
    3: "Tech"
}

ag_df['category'] = ag_df['label'].map(ag_label_map)
ag_df['text'] = ag_df['text']

ag_final = ag_df[['text', 'category']]
```

# 📤 🟢 CELL 3 — Upload BBC Dataset (Manual in Colab)

```python
bbc_df = pd.read_csv("/content/bbc-news-data.csv")

bbc_df = bbc_df[['category', 'text']]

bbc_map = {
    "business": "Business",
    "sport": "Sports",
    "tech": "Tech",
    "politics": "World",
    "entertainment": "World"
}

bbc_df['category'] = bbc_df['category'].map(bbc_map)

bbc_final = bbc_df[['text', 'category']]
```

# 🔗 🟢 CELL 4 — Merge Datasets

```python
df = pd.concat([ag_final, bbc_final], ignore_index=True)
```

# 📊 🟢 CELL 5 — EDA (Required)

```python
print(df.head())

print("\nDataset Info:")
print(df.info())

print("\nCategory Distribution:")
print(df['category'].value_counts())
```

# ⚠️ 🟢 CELL 6 — Check Missing Labels

```python
print("Missing categories:", df['category'].isnull().sum())
```

# 🔀 🟢 CELL 7 — Train/Test Split

```python
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)
```

# 💾 🟢 CELL 8 — Save Processed Data

```python
df.to_csv("full_dataset.csv", index=False)
train_df.to_csv("train.csv", index=False)
test_df.to_csv("test.csv", index=False)
```
